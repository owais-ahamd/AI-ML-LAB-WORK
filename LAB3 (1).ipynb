```python
# Step 1: Initialize variables
target_temp = int(input("Enter desired temperature: "))
environment_temp = int(input("Enter room temperature: "))

model = {"offset": 0}   # Agent knowledge
learning_rate = 0.1     # Learning speed

# Step 2: Run for 5 steps
for step in range(5):

    # Performance Element: Decision
    if environment_temp < target_temp:
        heater = "ON"
        applied_temp = environment_temp + 5 + model["offset"]
    else:
        heater = "OFF"
        applied_temp = environment_temp

    # Critic: Calculate error
    error = target_temp - applied_temp

    # Learning Element: Update knowledge
    model["offset"] += learning_rate * error

    # Problem Generator: Change environment
    if step % 2 == 0:
        environment_temp = applied_temp - 0.5
    else:
        environment_temp = applied_temp + 0.5

    # Output
    print(f"Step {step + 1}: Heater {heater}, Temp: {applied_temp:.1f}°C, Error: {error:.1f}, Model Offset: {model['offset']:.1f}")
```

```python
# Post-Lab-11 Task: Water-level learning agent 
import random

target_level = int(input("Enter target water level (litres): "))
current_level = int(input("Enter current water level (litres): "))

model = {"offset": 0}
learning_rate = 0.1

for step in range(10):

    # Performance Element: open valve if level is too low
    if current_level < target_level:
        valve = "OPEN"
        applied_level = current_level + 10 + model["offset"]
    else:
        valve = "CLOSED"
        applied_level = current_level

    # Critic
    error = target_level - applied_level

    # Learning Element
    model["offset"] += learning_rate * error

    # Problem Generator: random usage / rain event
    disturbance = random.choice([-1, 0, 1])
    current_level = applied_level + disturbance

    print(f"Step {step+1}: Valve {valve}, Level: {applied_level:.1f}L, Error: {error:.1f}, Offset: {model['offset']:.2f}")

```
