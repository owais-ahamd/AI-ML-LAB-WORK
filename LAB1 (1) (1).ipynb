```python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)

print(quicksort([3,6,8,10,1,2,1]))
```

```python
#Q8: Write a function which takes that numpy 1-D array as input and returns the following:
#1. Maximum value in the array
#2. Standard Deviation
#3. Sum of all the elements in the array
import numpy as np

def array_stats(x):
    max_val = np.max(x)
    std_val = np.std(x)
    sum_val = np.sum(x)
    
    return max_val, std_val, sum_val

# Example
x = np.array([2, 4, 6, 8, 10])

result = array_stats(x)

print("Max:", result[0])
print("Standard Deviation:", result[1])
print("Sum:", result[2])
```

```python
#Q9: Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.
def count_case(s):
    upper = 0
    lower = 0

    for ch in s:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1

    return upper, lower


# Example
text = "Hello World"

result = count_case(text)

print("Uppercase letters:", result[0])
print("Lowercase letters:", result[1])
```

```python
#Q11: Write a function, calculateSum(l,k) which will take a list, l and a number k as input and return two numbers from l that add up to k.
def calculateSum(l, k):
    for i in range(len(l)):
        for j in range(i + 1, len(l)):
            if l[i] + l[j] == k:
                return l[i], l[j]

    return None  # if no pair found


# Example
nums = [2, 7, 11, 15]
k = 9

result = calculateSum(nums, k)
print(result)
```

```python
class evenNumbers:
    def __init__(self, n):
        self.n = n
        self.current = 2   # first positive even number

    def __iter__(self):
        return self

    def __next__(self):   # Python 3 uses __next__
        if self.current <= self.n:
            num = self.current
            self.current += 2
            return num
        else:
            raise StopIteration

myrange = evenNumbers(10)

print(next(myrange))  # 2
print(next(myrange))  # 4
```

```python

import re

articles = [
    "Hello! This is article 1. یہ ایک مثال ہے۔",
    "AI @2026 is growing fast؛ amazing!",
    "Data 123 science - future of AI؟"
]

cleaned_articles = []

for article in articles:
    # 1. Remove punctuation (English + Urdu/Arabic)
    article = re.sub(r"[۔؟٫،؛٪/!@()°.\-;″′]", " ", article)

    # 2. Remove digits
    article = re.sub(r"\d+", " ", article)

    # 3. Remove single English characters (standalone letters)
    article = re.sub(r"\b[a-zA-Z]\b", " ", article)

    # 4. Remove extra spaces
    article = re.sub(r"\s+", " ", article).strip()

    cleaned_articles.append(article)

print("Cleaned Articles:")
print(cleaned_articles)
```

```python
from collections import Counter
import re

articles = [
    "میں نے موبائل استعمال کیا",
    "موبائل بہت مفید ہے",
    "مجھے معلومات حاصل کرنا پسند ہے",
    "موبائل اور میسجنگ اہم ہیں",
    "معلومات کا ذریعہ موبائل ہے"
]

words_list = []

for article in articles:
    # Remove punctuation (basic cleaning)
    article = re.sub(r"[۔؟٫،؛٪/!@()°.\-;″′]", " ", article)

    # Split into words
    words = article.split()

    # Collect words starting with "م"
    for word in words:
        if word.startswith("م"):
            words_list.append(word)

# Count frequency
word_counts = Counter(words_list)

# Top 10 most common
top_10 = word_counts.most_common(10)

print("Top 10 words starting with 'م':")
print(top_10)
```
