```python
def simple_vacuum():

    # Ask the user for the robot’s starting position  
    print("Welcome! Let’s clean two rooms!")

    location = input("Where is the robot? Type A or B: ").upper()
    while location not in ['A', 'B']:
        location = input("Oops! Type A or B: ").upper()

    # Ask about room dirtiness  
    room_a = int(input("Is Room A dirty? Type 1 for yes, 0 for no: "))
    while room_a not in [0, 1]:
        room_a = int(input("Oops! Type 0 or 1: "))

    room_b = int(input("Is Room B dirty? Type 1 for yes, 0 for no: "))
    while room_b not in [0, 1]:
        room_b = int(input("Oops! Type 0 or 1: "))

    # Track actions
    actions = []
    cost = 0

    print("\nLet’s start cleaning!")
    print(f"Robot is in Room {location}. Room A is {'dirty' if room_a else 'clean'}. Room B is {'dirty' if room_b else 'clean'}.")

    # Cleaning loop
    while room_a != 0 or room_b != 0:

        # Clean current room
        if (location == 'A' and room_a == 1):
            print("Cleaning Room A!")
            room_a = 0
            actions.append("Clean A")
            cost += 1

        elif (location == 'B' and room_b == 1):
            print("Cleaning Room B!")
            room_b = 0
            actions.append("Clean B")
            cost += 1

        # Move if current room is clean
        else:
            if location == 'A':
                print("Moving to Room B!")
                location = 'B'
                actions.append("Move Right")
            else:
                print("Moving to Room A!")
                location = 'A'
                actions.append("Move Left")
            cost += 1

    print("\nDone! Both rooms are clean.")
    print(f"Actions: {actions}")
    print(f"Total cost: {cost}")


# Run the program
simple_vacuum()
```
